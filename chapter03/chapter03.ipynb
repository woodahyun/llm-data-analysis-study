### 결과 관찰
세 가지 FK 관계에서 기준 파일에 존재하지 않는 ID가 모두 0건이었다.
- orders.customer_id는 모두 customers.customer_id에 존재한다.
- order_items.order_id는 모두 orders.order_id에 존재한다.
- order_items.product_id는 모두 products.product_id에 존재한다.

또한 order_items와 products를 product_id 기준으로 병합했을 때 764개 주문상세 행이 모두 상품 정보와 연결되었고, category 결측값도 발생하지 않았다. 따라서 현재 데이터에서는 주문상세별 금액인 quantity × unit_price를 계산하고 카테고리별로 집계할 수 있다.

### 나의 해석과 판단
- 키 관계 문제가 발견되었다고 해서 해당 행을 바로 삭제하면 안 된다. 연결되지 않은 ID는 단순 오류일 수도 있지만, 기준 파일이 아직 적재되지 않았거나 고객·상품 정보가 별도로 관리되는 업무상 이유 때문일 수도 있다.
- 삭제하면 어떤 주문이나 상품 정보가 누락되었는지 추적할 수 없고, 매출·주문 수·고객 수가 실제보다 작게 계산될 수 있다. 따라서 먼저 원본 데이터, 파일 생성 시점, 업무 규칙을 확인하고, 원인이 확인된 뒤 수정·보완·제외 여부를 결정해야 한다.

### 업무·분석적 의미
- CSV 간 키 관계 검증은 여러 파일을 결합한 분석의 신뢰성을 확인하는 기본 단계이다. 키가 연결되지 않으면 고객별 구매 분석, 주문별 매출 분석, 카테고리별 판매 분석에서 일부 데이터가 누락될 수 있다.
- 이번 데이터는 기본 FK 관계가 모두 유지되므로 고객-주문-주문상세-상품 정보를 연결해 분석할 수 있다. 즉, 고객 특성별 구매패턴이나 상품 카테고리별 매출을 계산할 수 있는 구조가 확인되었다.

### 한계와 추가 확인 사항
- 현재 검증은 “참조 대상 ID가 존재하는가”만 확인한 것으로, ID가 업무적으로 올바른 고객·주문·상품과 연결되었는지는 확인하지 못했다.
- 하나의 고객이 여러 주문을 하고, 하나의 주문에 여러 상품이 포함되는 관계가 예상한 구조와 정확히 일치하는지 추가 확인이 필요하다.
- 병합해 계산한 카테고리별 금액에는 취소·환불 주문도 포함될 수 있다. 실제 매출 분석에서는 orders.order_status = 'completed' 조건을 적용해야 한다.
- unit_price가 할인·쿠폰 등이 반영된 실제 판매단가인지, 상품 정가와 차이가 나는 이유는 추가 확인이 필요하다.

## 5. LLM 구조 설명 검증
- LLM에 제공한 Safe Context:
네 개 파일의 파일명, 행·열 수, 컬럼명, 데이터 타입, 결측치·중복 여부, 그리고 customers → orders → order_items ← products의 키 연결 관계를 제공했다. 원본 고객 이름이나 개별 주문 내용은 제공하지 않았다.

- LLM이 제안한 추가 점검:
age 컬럼의 실제 존재 여부와 결측·이상치 여부, customer_id 연결 여부, 매출 계산에 필요한 quantity와 unit_price 존재 여부, 완료·취소·환불 주문 처리 기준을 확인하도록 제안했다.

- 실제 데이터에서 확인한 항목:
customers에 age 컬럼이 존재하며 결측값은 없고, 값의 범위는 19~69세였다. orders.customer_id는 모두 customers.customer_id와 연결되었다. 또한 order_items에는 quantity와 unit_price가 모두 존재하며, 수량은 1~5개이고 음수·0 값은 없었다. 주문 상태는 completed, cancelled, refunded로 구성되어 있음을 확인했다.

- 채택/수정/보류한 내용:
채택: 컬럼 존재 여부, 결측치, 키 연결, 매출 계산 컬럼 확인  
수정: “연령대별 매출 분석을 바로 수행한다”는 제안을 완료 주문만 포함하고 quantity × unit_price로 매출을 계산하는 방식으로 구체화  
보류: 연령이 높거나 낮아서 매출이 달라진다는 원인 해석은 현재 데이터만으로 판단할 수 없어 보류

![LLM 구조 검토](images/step05_llm.png)

### 나의 해석과 판단
- 가장 유용했던 제안은 분석 전에 필요한 데이터 구조와 검증 항목을 체크리스트로 정리해 준 점이다. 특히 고객 정보만으로 매출을 분석할 수 없고, 주문·주문상세 데이터를 연결해 quantity × unit_price를 계산해야 한다는 점은 실제 분석 오류를 줄이는 데 도움이 된다.
- 가장 조심해야 할 부분은 LLM이 컬럼 하나의 존재만 보고 분석이 바로 가능하다고 판단할 수 있다는 점이다. 예를 들어 age 컬럼이 있어도 결측값, 이상치, 고객-주문 연결, 완료 주문 기준을 확인하지 않으면 연령대별 매출 분석 결과를 신뢰하기 어렵다.

### 한계와 추가 확인 사항
- 현재 확인은 데이터 구조와 기본 품질 중심이며, 연령대별 매출 차이가 통계적으로 의미 있는지는 아직 검증하지 않았다.
- age가 가입 시점 기준인지 주문 시점 기준인지 확인이 필요하다.
- unit_price가 할인·쿠폰이 반영된 실제 판매단가인지 확인해야 한다.
- 고객의 소득, 구매 목적, 할인 혜택 등 연령과 구매금액의 관계에 영향을 줄 수 있는 정보는 없다.
- LLM에 구조 요약만 제공했기 때문에, 실제 값의 세부 분포나 업무 규칙까지는 LLM이 알 수 없다는 한계가 있다.

## 6. Chapter 03 최종 판단
### 데이터의 첫인상 3가지
1. 고객·상품·주문·주문상세 데이터가 분리되어 있지만, customer_id, order_id, product_id를 통해 정상적으로 연결되는 관계형 구조이다.  
2. 주요 ID의 결측·중복과 파일 간 연결 누락이 없고, 숫자형 컬럼에도 0·음수처럼 명확한 오류가 보여 기본 분석을 시작할 수 있는 상태이다.  
3. 주문은 2025-07-09~2026-07-08의 약 1년 데이터이며, 결제수단은 비교적 고르게 분포하지만 상품 카테고리별 상품 수에는 차이가 있다.

### 다음 Chapter 전에 반드시 확인/처리해야 할 항목
1. order_status를 기준으로 완료 주문만 분석에 포함할지, 취소·환불 주문을 별도 지표로 볼지 기준을 정한다.  
2. order_items.quantity × order_items.unit_price로 주문상품 금액을 계산하고, 주문 단위 매출로 집계할 때 중복 계산이 없는지 확인한다.  
3. products.price와 order_items.unit_price의 차이가 할인·쿠폰·행사 가격에 따른 정상 차이인지 확인하고, 날짜 양 끝의 부분 월 데이터를 월별 분석에서 어떻게 처리할지 결정한다.

### 현재 데이터만으로 단정할 수 없는 것
현재 데이터만으로는 결제수단, 고객 연령, 성별, 상품 카테고리 중 어떤 요인이 소비금액을 변화시켰다고 단정할 수 없다. 또한 가격 차이가 할인 때문인지, 취소·환불이 발생한 이유가 무엇인지, 고객이 특정 상품이나 결제수단을 선택한 이유도 알 수 없다. 관찰된 차이는 데이터의 분포일 수 있으므로, 업무 규칙 확인과 추가 데이터·통계적 검증 없이 인과관계로 해석해서는 안 된다.
