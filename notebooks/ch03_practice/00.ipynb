```python
#전체적인 column에 대한 summary 
print(orders.info())

#column들만 갖고 오고 싶을 때
print(orders.columns)

#column들의 형(문자형 type, 숫자 등)을 보고 싶을 때
print(orders.dtypes)
```

```python
#특정 column 하나만 볼 때
print(orders["payment_method"].isna().sum())

#전체 coluumn의 결측치를 볼 때
print(orders.isna().sum())
```

```python
id_missing = orders["customer_id"].isna().sum()
id_duplicates = orders["customer_id"].duplicated().sum()
order_duplicates = orders["order_id"].duplicated().sum()

print("ID 결측:", id_missing)
print("ID 중복:", id_duplicates)
print("Order 중복:", order_duplicates)
```
